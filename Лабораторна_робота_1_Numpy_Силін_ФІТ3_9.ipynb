### Завдання 28
Для матриці:

```text

     [2, 6]]
```

обчисліть:

````python
# Код студента
A = np.array([[4, 7], [2, 6]])
print(np.linalg.det(A))
A_inv = np.linalg.inv(A)
print(A_inv)
print(A @ A_inv)  # має вийти одинична матриця
print(np.allclose(A @ A_inv, np.eye(2)))
````

### Завдання 33
Нормалізуйте **кожен стовпець** матриці:

```text
A = [[10, 20, 30],
     [20, 30, 40],
     [30, 40, 50]]
```

за формулою:

\[
z = \frac{x - \mu}{\sigma}
\]

де `μ` — середнє значення стовпця, `σ` — стандартне відхилення стовпця.

Не використовуйте цикл `for`.

````python
# Код студента
A = np.array([[10, 20, 30], [20, 30, 40], [30, 40, 50]])
Z = (A - A.mean(axis=0)) / A.std(axis=0)
print(Z)
````

````python
# Код студента
M = np.random.randint(1, 21, (5, 5))
print(M)
print("головна діагональ:", np.trace(M))
print("побічна діагональ:", np.trace(np.fliplr(M)))
print("max рядків:", M.max(axis=1))
print("min стовпців:", M.min(axis=0))
M[M < M.mean()] = 0
print(M)
````

### Завдання 40
Створіть матрицю коефіцієнтів і вектор правих частин для системи:

```text
 x + 2y +  z =  9
2x +  y -  z =  8
3x -  y + 2z =  3
```

1. Розв’яжіть систему за допомогою `np.linalg.solve()`.
2. Перевірте знайдений розв’язок матричним множенням `A @ x`.
